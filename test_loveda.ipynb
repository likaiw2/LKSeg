```python
from PIL import Image
import numpy as np
import os

f_path = '/home/likai/code/LKSeg/data/LoveDA/Train/Rural/masks_png'
all_values = set()  # 使用 set 来自动去重

for img_name in os.listdir(f_path):
    image = Image.open(f'{f_path}/{img_name}')
    image_np = np.array(image)
    uni = np.unique(image_np)
    all_values.update(uni)  # 加入 set 中，自动去重

# 转成排序后的 tuple
unique_values = tuple(sorted(all_values))
print(unique_values)
```
