```python
from data_reader.loveda_dataset import LoveDATrainDataset
from torch.utils.data import DataLoader

train_dataset = LoveDATrainDataset(data_root='data/LoveDA/Train')
train_loader = DataLoader(
    train_dataset,
    batch_size=1,
    num_workers=2,
    pin_memory=True,
    shuffle=True,
    drop_last=True
)

print("########")
sample = next(iter(train_loader))
print(sample['img'].size())
print(sample['gt_semantic_seg'].size(), sample['gt_semantic_seg'].unique())
print(sample['img_id'])
print(sample['img_type'])

import matplotlib.pyplot as plt

plt.imshow(sample['img'][0].permute(1, 2, 0))
plt.show()

plt.imshow(sample['gt_semantic_seg'][0])
plt.show()

print("########")
```
